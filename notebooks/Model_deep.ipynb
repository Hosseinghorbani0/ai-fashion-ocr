```python
import tensorflow as tf
from keras.models import Sequential 
from keras.layers import   MaxPooling2D , Flatten , Dense , Conv2D ,Dropout
import sklearn.metrics as sm
import cv2
import numpy as np
import os
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

%matplotlib inline

```

```python
x=np.empty((0,1024))


for s in range (1,13):
 files=os.listdir("ax/"+str(s))
 for filename in files:
     full_path = os.path.join("ax/"+str(s), filename) 
     im = cv2.imread(full_path)
     im2=cv2.resize(im,(32,32))
     im3=cv2.cvtColor(im2,cv2.COLOR_BGR2GRAY)
     im4=im3.flatten()
     im5=im4/255
     x=np.append(x,[im5],axis=0)
```

```python
encoder = LabelEncoder()
y_gheymat=np.array([4500000,4500000,3100000,3100000,3100000,3800000,3800000,3800000,2500000,2500000,4200000,4200000,4200000,2100000,2100000,3200000,3200000,3200000,3200000,3600000,3600000,3600000,3600000,3600000,3600000,4100000,4100000,1500000,1500000,2200000,2200000,2200000,3400000,3400000,3400000])
y_gheymat = encoder.fit_transform(y_gheymat)
```

```python
encoder = LabelEncoder()
y_size=np.array([42,42,46,46,46,48,48,48,50,50,46,46,46,44,44,50,50,50,50,44,44,44,44,44,44,50,50,42,42,48,48,48,44,44,44])
y_size= encoder.fit_transform(y_size)
```

```python
encoder = LabelEncoder()
y_ghad=np.array([70,70,70,70,70,80,80,80,80,80,75,75,75,75,75,80,80,80,80,70,70,70,70,70,70,100,100,80,80,75,75,75,75,75,75])
y_ghad= encoder.fit_transform(y_ghad)
```

```python
encoder = LabelEncoder()
y_jense = np.array([
    "کتان استر دار", "کتان استر دار", "فوتر", "فوتر", "فوتر",
    "ژاکاد", "ژاکارد", "ژاکارد", "فوتر", "فوتر", "فوتر", "فوتر",
    "فوتر", "فوتر", "فوتر", "فوتر", "فوتر", "فوتر", "فوتر",
    "فوتر آستردار", "فوتر آستردار", "فوتر آستردار", "فوتر آستردار",
    "فوتر آستردار", "فوتر آستردار", "فوتر آستردار", "فوتر آستردار",
    "فوتر و چرم و جین", "فوتر و چرم و جین", "فوتر", "فوتر",
    "فوتر", "فوتر", "فوتر", "فوتر"
])

y_jense= encoder.fit_transform(y_jense)
```

```python
x_Train1,x_Test1,y_Train1,y_Test1=train_test_split(x,y_gheymat,test_size=0.001)
x_Train2,x_Test2,y_Train2,y_Test2=train_test_split(x,y_size,test_size=0.001)
x_Train3,x_Test3,y_Train3,y_Test3=train_test_split(x,y_ghad,test_size=0.001)
x_Train4,x_Test4,y_Train4,y_Test4=train_test_split(x,y_jense,test_size=0.001)
```

```python
y_Train4.shape
```

```python
x_Train1 = x_Train1.reshape((-1, 32, 32, 1)) 
x_Test1 = x_Test1.reshape((-1, 32, 32, 1))

x_Train2 = x_Train2.reshape((-1, 32, 32, 1)) 
x_Test2 = x_Test2.reshape((-1, 32, 32, 1))

x_Train3 = x_Train3.reshape((-1, 32, 32, 1)) 
x_Test3 = x_Test3.reshape((-1, 32, 32, 1))

x_Train4 = x_Train4.reshape((-1, 32, 32, 1)) 
x_Test4 = x_Test4.reshape((-1, 32, 32, 1))
```

```python
model = Sequential()
model.add(Conv2D(128, kernel_size=(3, 3), activation="relu", input_shape=(32, 32, 1), padding="same"))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Dropout(0.3))
model.add(Conv2D(256, kernel_size=(3, 3), activation="relu", padding="same"))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Dropout(0.3))
model.add(Conv2D(320, kernel_size=(3, 3), activation="relu", padding="same"))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Dropout(0.3))
model.add(Conv2D(256, kernel_size=(3, 3), activation="relu", padding="same"))
model.add(Flatten())
model.add(Dense(128, activation="relu"))
model.add(Dropout(0.3))
model.add(Dense(35, activation="softmax"))

model.summary()
```

```python
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=['accuracy'])
h=model.fit(x_Train1, y_Train1, batch_size=32,epochs=500,validation_data=(x_Test1,y_Test1))
```

```python
# ارزیابی مدل روی داده‌های تست
train_loss, train_accuracy = model.evaluate(x_Train1, y_Train1)
print(f"train Accuracy: {train_accuracy * 100:.2f}%")
print(f"train Loss: {train_loss:.4f}")
```
